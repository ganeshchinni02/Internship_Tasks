Q1. Find the Frequency of Each Word in a Sentence
1. Start the program.
2. Take a sentence as input.
3. Create an empty dictionary called frequency.
4. Convert the sentence to lowercase and split it into words.
5. Check each word in the sentence.
6. If the word already exists in the dictionary, increase its count by 1.
7. Otherwise, add the word to the dictionary with count 1.
8. Return the frequency dictionary.
9. Stop the program.

```python
def word_frequency(sentence):
    frequency = {}
    for word in sentence.lower().split():
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    return frequency
print(word_frequency("python is easy and python is powerful"))
print(word_frequency("M.S Dhoni is a great cricketer and M.S Dhoni is a great captain"))
```

Q2. Find the Frequency of Each Character in a Word
1. Start the program.
2. Take a word as input.
3. Create an empty dictionary.
4. Check each character in the word.
5. If the character is already in the dictionary, add 1 to its count.
6. If the character is new, store it with count 1.
7. Return the dictionary.

```python
def character_frequency(word):
    frequency = {}
    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    return frequency

print(character_frequency("Mahendra Singh Dhoni"))
print(character_frequency("bhuvneshwar kumar"))
```

Q3. Find the First Non-Repeating Character in a Word
Pseudocode (Simple English)
1. Start the program.
2. Take a word as input.
3. Check each character in the word.
4. Count how many times the character appears.
5. If the character appears only once, return that character.
6. If no character appears only once, return None.

```python
def first_non_repeating_character(word):
    for char in word:
        if word.count(char) == 1:
            return char

    return None

print(first_non_repeating_character("cover drive"))
print(first_non_repeating_character("helicopter"))
```

Q4. Check Whether Two Strings Are Anagrams
Pseudocode (Simple English)
1. Start the program.
2. Take two strings as input.
3. Convert both strings to lowercase.
4. Sort the characters in both strings.
5. Compare the sorted strings.
6. If both strings are equal, return True.
7. Otherwise, return False.

```python
def are_anagrams(s1, s2):
    s1 = s1.lower()
    s2 = s2.lower()

    if sorted(s1) == sorted(s2):
        return True
    else:
        return False

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))    
```

Q5. Find the First Repeating Character in a String
1. Start the program.

3. Create an empty set to store characters.
4. Check each character in the string from left to right.
5. If the character is already in the set, return that character.
6. Otherwise, add the character to the set.
7. If no character repeats, return None.

Q6. Move All Zeros to the End of a List
1. Start the program.
2. Take a list as input.
3. Create a new list with all non-zero elements.
4. Count the number of zeros in the original list.
5. Add all zeros to the end of the new list.
6. Return the new list.

```python
def move_zeros_to_end(numbers):
    result = []

    for num in numbers:
        if num != 0:
            result.append(num)

    for num in numbers:
        if num == 0:
            result.append(num)

    return result

numbers = [0, 1, 2, 0, 3, 4, 0, 5]
numbers1 = [0, 0, 0, 1, 2, 3]
print(move_zeros_to_end(numbers))
print(move_zeros_to_end(numbers1))
```

Q7. Find the Second Highest Distinct Number in a List
1. Start the program.
2. Take a list of numbers as input.
3. Remove duplicate numbers from the list.
4. Sort the numbers in descending order.
5. If at least two distinct numbers exist, return the second number.
6. Otherwise, return None.

```python
def second_highest_distinct(numbers):
    unique = list(set(numbers))
    unique.sort(reverse=True)

    if len(unique) >= 2:
        return unique[1]
    else:
        return None

numbers = [10, 20, 30, 40, 50]
numbers1 = [5, 10, 15, 20, 25]
print(second_highest_distinct(numbers))  
print(second_highest_distinct(numbers1))    
```

Q8. Remove Duplicates Without Changing the Original Order
1. Start the program.
2. Take a list as input.
3. Create an empty list to store unique elements.
4. Check each element in the original list.
5. If the element is not in the new list, add it.
6. Return the new list.

```python
def remove_duplicates_preserve_order(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result

numbers = [1, 2, 3, 2, 4, 3, 5]
numbers1 = [10, 20, 30, 20, 40, 30, 50]
print(remove_duplicates_preserve_order(numbers))
print(remove_duplicates_preserve_order(numbers1))
```

Q9. Find the Missing Number from 1 to N
1. Start the program.
2. Take a list of numbers and the value of N as input.
3. Find the sum of all numbers from 1 to N.
4. Find the sum of the numbers in the given list.
5. Subtract the list sum from the total sum.
6. Return the missing number.

```python
def find_missing_number(nums, n):
    for num in range(1, n + 1):
        if num not in nums:
            return num
    return None
numbers = [1, 2, 4, 5, 6]
numbers1 = [1, 2, 3, 5, 6]
print(find_missing_number(numbers, 6))
print(find_missing_number(numbers1, 6))
```

Q10. Find Two Numbers Whose Sum Equals a Target
1. Start the program.
2. Take a list of numbers and a target value as input.
3. Check each number in the list from left to right.
4. For each number, check the numbers before it.
5. If two numbers add up to the target, return them as a list.
6. If no pair is found, return an empty list.

```python
def two_sum_first_pair(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []

numbers = [2, 7, 11, 15]
numbers1 = [3, 2, 4]
target = 9
print(two_sum_first_pair(numbers, target))
print(two_sum_first_pair(numbers1, target))
```

Q11. Find the Number Closest to Zero
1. Start the program.
2. Take a list of numbers as input.
3. Find the number closest to zero.
4. If two numbers are equally close to zero, choose the positive number.
5. Return the result.

```python
def closest_to_zero(nums):
    closest = nums[0]

    for num in nums:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num

    return closest
numbers = [3, -2, 5, -1, 0, 4]
numbers1 = [-5, -3, -1, 2, 4]   
print(closest_to_zero(numbers))
print(closest_to_zero(numbers1))
```

Q12. Find Common Elements Between Two Lists
1. Start the program.
2. Take two lists as input.
3. Create an empty list to store common elements.
4. Check each element in the first list.
5. If the element is present in the second list and not already in the result, add it.
6. Return the result.

```python
def common_elements(list1, list2):
    result = []

    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
list3 = ['apple', 'banana', 'cherry']           
list4 = ['banana', 'kiwi', 'grape']
print(common_elements(list1, list2))
print(common_elements(list3, list4))
```

Q14. Find the Maximum Sum of Two Consecutive Elements
1. Start the program.
2. Take a list of numbers as input.
3. Add the first two consecutive numbers.
4. Store their sum as the maximum sum.
5. Check each pair of consecutive numbers in the list.
6. If a pair has a larger sum, update the maximum sum.
7. Return the maximum sum.

```python
def max_consecutive_pair_sum(nums):
    maximum = nums[0] + nums[1]

    for i in range(len(nums) - 1):
        total = nums[i] + nums[i + 1]

        if total > maximum:
            maximum = total

    return maximum
numbers = [1, 2, 3, 4, 5]
numbers1 = [10, 20, 30, 40, 50]
print(max_consecutive_pair_sum(numbers))
print(max_consecutive_pair_sum(numbers1))
```

Q15. Find All Pairs of Numbers Whose Sum Equals a Target
1. Start the program.
2. Take a list of numbers and a target value as input.
3. Create an empty list to store pairs.
4. Check each pair of numbers in the list.
5. If their sum equals the target, arrange the pair in ascending order.
6. Add the pair if it is not already in the result.
7. Sort the result by the first element of each pair.
8. Return the result.

```python
def all_unique_pairs(nums, target):
    result = []

    for i in nums:
        for j in nums:
            if i < j and i + j == target:
                if [i, j] not in result:
                    result.append([i, j])

    result.sort()
    return result
numbers = [1, 2, 3, 4, 5]
numbers1 = [2, 4, 6, 8, 10]
print(all_unique_pairs(numbers, 5))
print(all_unique_pairs(numbers1, 10))
```
